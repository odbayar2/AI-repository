### Please define X and Y to predict `class` through `wine_df`.

```python
import pandas as pd
from sklearn.datasets import load_wine

wine = load_wine()

wine_df = pd.DataFrame(wine['data'], columns=wine.feature_names)
wine_df['class'] = wine.target

# Setting X, Y
X = wine_df.drop('class', axis=1)
Y = wine_df['class']
```

```python
print(X.head())
```

```python
print(Y.head())
print(Y.tail())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=42)

# Scaling (after splitting!)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled Training Set Sample:\n", X_train_scaled[:5])
print("Scaled Test Set Sample:\n", X_test_scaled[:5])
```

### Split train and test data through `train_test_split`.
- test_size = 0.3, random_state = 0

```python
from sklearn.model_selection import train_test_split

# Splitting Train/Test data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=0)
```

```python
# Scaling (MUST be done after splitting — to avoid data leakage)
scaler = StandardScaler()

# Fit on training data only
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using the scaler fitted on train data
X_test_scaled = scaler.transform(X_test)

print("Scaled X_train sample:")
print(X_train_scaled[:5])

print("\nScaled X_test sample:")
print(X_test_scaled[:5])
```

##Build Models Using Various Models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

# Generate Model (Example: Logistic Regression)
model = LogisticRegression(max_iter=1000)

# Fitting
model.fit(X_train_scaled, Y_train)

# Get predict value by predicting with test data
Y_predict = model.predict(X_test_scaled)
```

##Check the fitted model's performance


```python
from sklearn.metrics import classification_report
print(classification_report(Y_test, Y_predict))
```

### Print the Accuracy.

```python
# model accuarcy
accuracy = model.score(X_test_scaled, Y_test)
print("Accuracy: {0:.3f}".format(accuracy))
```
